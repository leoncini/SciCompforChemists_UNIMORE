Ogni capitolo è un file presente sul repository. Si tratta di file di tipo <span style="color:blue">jupyter notebooks</span>, file cioè in cui è possibile inserire **testo e grafica**, come ad esempio in un file MS Word, ma anche **codice Python** che può essere eseguito, se il notebook è collegato ad un interprete del linguaggio (detto <span style="color:blue">kernel</span>). I notebook sono file con estensione **ipynb**.

Per poter utilmente visulizzare un file ipynb ed eseguire il codice python in esso (eventualmente) contenuto, è necessario avere installato software opportuno. Le macchine del laboratorio hanno installato tale software. Tuttavia, si tratta di macchine che possiamo etichettare come **computer senza memoria**. Questo vuol dire che allo spegnimento, il lavoro svolto (ovvero le eventuali modifiche apportate al/ai notebook) viene perrduto. Non solo per poter svolgere esercizi a casa, ma anche per le lezioni in laboratorio è dunque consigliabile che ogni studente utilizzi il proprio **computer portatile, con il software necessario.

Il modo più semplice di procedere è di installare un unico pacchetto software, chiamato <span style="color:blue">Anaconda</span>. Nel notebook <span style="color:blue">chap__00_notebook.ipynb</span> sono riportate istruzioni dettagliate. Tuttavia, per aprire questo notebook con le istruzioni per istallare Anaconda è necessario disporre di Anaconda... Un problema circolare. Per questa ragione, nel repository è comunque presente la versione html di tale notebook.
